```python
import inspect
import pathlib  # Example package to inspect

def list_package_contents(package):
    print(f"=== Inspecting Package: {package.__name__} ===\n")
    
    # 1. List Sub-modules (if it's a parent package)
    print("📁 SUB-MODULES:")
    for name, obj in inspect.getmembers(package, inspect.ismodule):
        if not name.startswith("_"):  # Skip private items
            print(f"  - {name}")
            
    # 2. List Classes
    print("\n🏛️ CLASSES:")
    for name, obj in inspect.getmembers(package, inspect.isclass):
        if not name.startswith("_"):
            print(f"  - {name}")
            
    # 3. List Functions
    print("\n⚙️ FUNCTIONS / METHODS:")
    for name, obj in inspect.getmembers(package, inspect.isfunction):
        if not name.startswith("_"):
            print(f"  - {name}")
            
    # 4. List Variables, Constants, and Other Properties
    print("\n📌 VARIABLES & CONSTANTS:")
    for name in dir(package):
        if not name.startswith("_"):
            obj = getattr(package, name)
            # Filter out things that are modules, classes, or functions
            if not (inspect.ismodule(obj) or inspect.isclass(obj) or inspect.isfunction(obj)):
                print(f"  - {name} ({type(obj).__name__}): {obj}")

# Run the inspector on the pathlib module
list_package_contents(pathlib)
```
